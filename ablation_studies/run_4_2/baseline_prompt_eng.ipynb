```python
# ── 2b. Prompt template ───────────────────────────────────────────────────────
CHOICE_LETTERS = "ABCDEFGHIJ"

# CODE CHANGE: build_prompt now accepts `processor` as a required argument
# so it can call processor.apply_chat_template. This means the datasets must
# be instantiated AFTER the processor is loaded (done in cell 4c below).
def build_prompt(row: pd.Series, proc, include_answer: bool = False) -> str:
    """
    CODE CHANGE: Build an instruct chat-template formatted prompt.

    Previously: raw string "<image>\nContext:...\nAnswer: B"
    Now:        proper <|im_start|>user / <|im_start|>assistant format
                produced by processor.apply_chat_template.

    WHY (prompt structure):
    SmolVLM-500M-Instruct was fine-tuned with a specific chat template.
    Using the raw completion format is misaligned with its training
    distribution. apply_chat_template ensures the prompt matches exactly
    what the model saw during instruct fine-tuning, giving a stronger
    prior before any LoRA adaptation.

    WHY (full answer text vs single letter):
    When include_answer=True (training), the correct choice's FULL TEXT
    is appended to the assistant turn instead of just the letter 'B'.
    This provides N tokens of training signal per example (where N is
    the length of the answer text) instead of just 1, giving far richer
    gradients. The model also learns to reason semantically rather than
    memorising an arbitrary letter-to-position mapping.

    Parameters
    ----------
    row            : a single DataFrame row (pd.Series)
    proc           : the loaded AutoProcessor (needed for apply_chat_template)
    include_answer : if True, append the full ground-truth choice text
                     in the assistant turn (for training only)
    """
    context_parts = []
    lecture = row.get("lecture", "")
    hint    = row.get("hint", "")
    if pd.notna(lecture) and str(lecture).strip():
        context_parts.append(str(lecture).strip())
    if pd.notna(hint) and str(hint).strip():
        context_parts.append(str(hint).strip())
    context_str = "\n".join(context_parts)

    # CODE CHANGE: A/B/C/D labels are kept in the choices list so the model
    # understands the multiple-choice structure, but the output it produces
    # (and the text we score at inference) is the full choice text, not the letter.
    choices_str = "\n".join(
        f"  {CHOICE_LETTERS[i]}. {c}" for i, c in enumerate(row["choices"])
    )

    # Build the user turn body (image placeholder goes in the message dict below)
    user_text = ""
    if context_str:
        user_text += f"Context:\n{context_str}\n\n"
    user_text += f"Question: {row['question']}\n"
    user_text += f"Choices:\n{choices_str}"

    # CODE CHANGE: structure prompt as a proper chat message with an explicit
    # image content entry. The processor maps {"type": "image"} to the correct
    # image token sequence when processor() is called with images=.
    messages = [
        {
            "role": "user",
            "content": [
                {"type": "image"},
                {"type": "text", "text": user_text},
            ],
        }
    ]

    # CODE CHANGE: apply_chat_template produces the <|im_start|>/<|im_end|>
    # formatted string aligned with SmolVLM-Instruct's training distribution.
    # add_generation_prompt=True appends "<|im_start|>assistant\n" so the
    # model knows to generate the answer next, or so we can append and score
    # candidate answer texts.
    # tokenize=False returns a plain string (tokenisation happens in processor()).
    prompt = proc.apply_chat_template(
        messages,
        add_generation_prompt=True,
        tokenize=False,
    )

    if include_answer:
        answer_idx  = int(row["answer"])
        # CODE CHANGE: append the full choice TEXT (e.g. "West Virginia"),
        # not just the letter (e.g. " A"). Followed by eos_token to close
        # the assistant turn cleanly.
        answer_text = row["choices"][answer_idx]
        prompt += answer_text + proc.tokenizer.eos_token

    return prompt


# CODE CHANGE: helper that returns ONLY the correct choice text string.
# Used by train_collate to build labels and by score_choices_batch to
# identify prompt boundaries during length-normalised scoring.
def get_answer_text(row: pd.Series) -> str:
    """Return the full text of the ground-truth answer choice."""
    return row["choices"][int(row["answer"])]


# NOTE: sanity-check prints are in cell 4c (after processor is loaded)
print("build_prompt and get_answer_text defined — sanity check runs after processor load.")
```

```python
# ── 2c. PyTorch Dataset ───────────────────────────────────────────────────────
class ScienceQADataset(Dataset):
    """
    Loads images from disk and builds prompts on-the-fly.

    CODE CHANGE: now accepts `proc` (the AutoProcessor) so build_prompt
    can call apply_chat_template. Datasets must therefore be instantiated
    AFTER the processor is loaded — see cell 4c.

    is_train=True  → returns chat-formatted prompt (no answer) + answer_text
                     separately, so train_collate can mask labels precisely.
    is_train=False → returns chat-formatted prompt + raw choices list
                     for length-normalised MC scoring.
    """

    def __init__(
        self,
        df: pd.DataFrame,
        data_dir: Path,
        proc,               # CODE CHANGE: processor needed for build_prompt
        img_size: int = 224,
        is_train: bool = True,
    ):
        self.df        = df.reset_index(drop=True)
        self.data_dir  = data_dir
        self.proc      = proc   # CODE CHANGE: store processor reference
        self.img_size  = img_size
        self.is_train  = is_train

    def __len__(self) -> int:
        return len(self.df)

    def _load_image(self, rel_path: str) -> Image.Image:
        full_path = self.data_dir / rel_path
        img = Image.open(full_path).convert("RGB")
        img = img.resize((self.img_size, self.img_size), Image.BICUBIC)
        return img

    def __getitem__(self, idx: int) -> dict:
        row = self.df.iloc[idx]
        img = self._load_image(row["image_path"])

        if self.is_train:
            return {
                "image":       img,
                # CODE CHANGE: prompt WITHOUT answer (include_answer=False).
                # train_collate concatenates answer_text separately so it can
                # compute the exact prompt token boundary for label masking.
                "text":        build_prompt(row, self.proc, include_answer=False),
                # CODE CHANGE: full answer text string (e.g. "West Virginia")
                # instead of the letter index. train_collate appends this to
                # build the full training sequence.
                "answer_text": get_answer_text(row),
                "answer":      int(row["answer"]),
            }
        else:
            return {
                "image":   img,
                "text":    build_prompt(row, self.proc, include_answer=False),
                "choices": row["choices"],
                "id":      row["id"],
                "answer":  int(row["answer"]) if "answer" in row else -1,
            }


# CODE CHANGE: dataset instantiation has been moved to cell 4c (below the
# processor load) because ScienceQADataset now requires `proc`. Running this
# cell alone will only define the class; the instances train_ds/val_ds/test_ds
# are created after the processor is available.
print("ScienceQADataset class defined — instances created after processor load (cell 4c).")
```

## 3. Model + QLoRA

```python
# ── 3a. Processor ─────────────────────────────────────────────────────────────
processor = AutoProcessor.from_pretrained(MODEL_ID)
processor.image_processor.max_image_tiles = 1  # disable splitting; 1 tile = base patches only
processor.image_processor.do_image_splitting = False

# Ensure pad token is defined (required for batched inference)
if processor.tokenizer.pad_token is None:
    processor.tokenizer.pad_token = processor.tokenizer.eos_token

print(f"Vocab size : {processor.tokenizer.vocab_size:,}")
print(f"Pad token  : '{processor.tokenizer.pad_token}'")
```

```python
# ── 3b. 4-bit Quantization Config (QLoRA) ────────────────────────────────────
# NF4 quantization keeps the backbone frozen in 4-bit;
# only the LoRA adapter weights (float16) are trained.
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",          # Normal-Float 4 — best for LLMs
    bnb_4bit_use_double_quant=True,     # Nested quantization saves ~0.4 bit/param
    bnb_4bit_compute_dtype=bnb_compute_dtype,
)

print("BitsAndBytes config ready.")
```

```python
# ── 3c. Load Pretrained HF Model ─────────────────────────────────────────────────────────────
model = AutoModelForImageTextToText.from_pretrained(
    MODEL_ID,
    quantization_config=bnb_config if use_bnb else None,
    dtype=model_load_dtype,
    device_map="auto",
    low_cpu_mem_usage=True,
)

if not torch.cuda.is_available():
    model = model.to(device)

# Disable caching — incompatible with gradient checkpointing
model.config.use_cache = False

print("Base model loaded.")
```

## 4. Collate Functions

```python
# ── 4a. Training collate ──────────────────────────────────────────────────────
# The full sequence (prompt + answer text) is fed in.
# Labels = input_ids with prompt tokens masked to -100 so loss is computed
# only on the answer text tokens.

def train_collate(batch: list[dict]) -> dict:
    images       = [item["image"]       for item in batch]
    prompt_texts = [item["text"]        for item in batch]  # prompt only
    answer_texts = [item["answer_text"] for item in batch]  # CODE CHANGE: full answer text

    # CODE CHANGE: build the full training sequence by concatenating the
    # chat-formatted prompt with the full answer text + eos_token.
    # Previously: prompt already contained " B" (the letter); the collate
    #             stripped the letter back off to find the boundary.
    # Now:        prompt and answer are kept separate until here, so the
    #             boundary is exact with no string-searching heuristics.
    eos = processor.tokenizer.eos_token
    full_texts = [p + a + eos for p, a in zip(prompt_texts, answer_texts)]

    # Tokenise the full sequences (prompt + answer text)
    encoding = processor(
        text=full_texts,
        images=images,
        return_tensors="pt",
        padding=True,
        truncation=True,
        max_length=MAX_SEQ_LEN,
    )

    input_ids      = encoding["input_ids"]      # (B, L)
    attention_mask = encoding["attention_mask"]  # (B, L)

    # ── Build labels: mask prompt tokens, supervise only on answer text ────────
    # Tokenise just the prompt (no answer) to find the exact prompt length.
    # CODE CHANGE: previously the prompt was recovered by rfind(" B") which
    # was fragile (could match the letter elsewhere in the text). Now we have
    # the clean prompt string directly from the dataset item.
    prompt_enc = processor(
        text=prompt_texts,
        images=images,
        return_tensors="pt",
        padding=True,
        truncation=True,
        max_length=MAX_SEQ_LEN,
    )
    prompt_lens = prompt_enc["attention_mask"].sum(dim=1)  # (B,) actual token counts

    labels = input_ids.clone()
    for i, plen in enumerate(prompt_lens):
        labels[i, :plen] = -100   # mask all prompt tokens
    labels[attention_mask == 0] = -100  # mask padding

    encoding["labels"] = labels
    return encoding


# ── 4b. Eval collate ─────────────────────────────────────────────────────────
# Returns the raw batch dict — MC scorer processes each sample individually.
# No change needed here: choices are already the full text strings.
def eval_collate(batch: list[dict]) -> dict:
    return {
        "images":   [item["image"]   for item in batch],
        "texts":    [item["text"]    for item in batch],
        "choices":  [item["choices"] for item in batch],
        "ids":      [item["id"]      for item in batch],
        "answers":  [item["answer"]  for item in batch],
    }


# ── 4c. Dataset instantiation (CODE CHANGE: moved here from cell 2c) ─────────
# Datasets are created here — after the processor is loaded — because
# ScienceQADataset now requires `processor` to call apply_chat_template.
train_ds = ScienceQADataset(train_df, DATA_DIR, processor, IMG_SIZE, is_train=True)
val_ds   = ScienceQADataset(val_df,   DATA_DIR, processor, IMG_SIZE, is_train=False)
test_ds  = ScienceQADataset(test_df,  DATA_DIR, processor, IMG_SIZE, is_train=False)

print(f"train_ds : {len(train_ds):,}")
print(f"val_ds   : {len(val_ds):,}")
print(f"test_ds  : {len(test_ds):,}")

# CODE CHANGE: sanity-check prints moved here (processor now available)
print("\n=== SAMPLE TRAINING PROMPT (chat template, full answer text) ===")
print(build_prompt(train_df.iloc[0], processor, include_answer=True))
print("\n=== SAMPLE INFERENCE PROMPT (chat template, no answer) ===")
print(build_prompt(val_df.iloc[0], processor, include_answer=False))


# ── 4d. DataLoaders ───────────────────────────────────────────────────────────
train_loader = DataLoader(
    train_ds,
    batch_size=TRAIN_BATCH,
    shuffle=True,
    collate_fn=train_collate,
    num_workers=2,
    pin_memory=True,
)

val_loader = DataLoader(
    val_ds,
    batch_size=EVAL_BATCH,
    shuffle=False,
    collate_fn=eval_collate,
    num_workers=2,
    pin_memory=True,
)

test_loader = DataLoader(
    test_ds,
    batch_size=EVAL_BATCH,
    shuffle=False,
    collate_fn=eval_collate,
    num_workers=2,
    pin_memory=True,
)

print(f"\nTrain batches : {len(train_loader):,}")
print(f"Val batches   : {len(val_loader):,}")
print(f"Test batches  : {len(test_loader):,}")
```

## 5. Multiple-Choice Log-Likelihood Scoring

Instead of generating tokens greedily, we score each candidate answer by computing its
**length-normalised log-likelihood** under the model.

**CODE CHANGE — two improvements over the original:**
1. **Full answer text** is scored (e.g. `"West Virginia"`) instead of a single letter token (`" A"`),
   giving richer, semantically meaningful comparisons between candidates.
2. **Length normalisation** divides each candidate's total log-prob by its token count,
   preventing the model from systematically favouring shorter answers due to probability accumulation.

This is still **exact, deterministic, and fast** — no beam-search needed.


```python
# ── 5. Batched multiple-choice scoring ───────────────────────────────────────

@torch.inference_mode()
def score_choices_batch(
    model,
    proc,
    images: list,
    prompt_texts: list[str],
    choices_list: list[list[str]],
) -> list[int]:
    """
    For each (image, prompt, choices) triple, score every candidate answer
    by its length-normalised log-likelihood and return the argmax.

    CODE CHANGE — Algorithm vs original:
    ─────────────────────────────────────
    OLD: full_seq = prompt + " A"   (single letter, 1 token scored)
         score   = sum of log-probs (unnormalised)

    NEW: full_seq = prompt + choice_text   (full answer text, N tokens scored)
         score   = sum of log-probs / N    (length-normalised)

    WHY full text:
      Scoring the complete choice string means the model's full language
      modelling capability is used to evaluate each candidate semantically,
      not just whether an arbitrary letter token follows the prompt.

    WHY length normalisation:
      Without it, shorter choices (e.g. "Yes", 1 token) accumulate less
      log-probability mass than longer ones (e.g. "West Virginia", 2 tokens)
      even if they are equally likely per token. Dividing by token count
      gives a per-token average that is comparable across choices of any length.

    Parameters
    ----------
    model         : PEFT-wrapped VLM (eval mode)
    proc          : matching AutoProcessor
    images        : list of PIL images  (length = B)
    prompt_texts  : list of chat-formatted prompt strings  (length = B)
    choices_list  : list of choice-string lists  (length = B)

    Returns
    -------
    preds : list[int]  (length = B)
    """
    model.eval()

    all_texts   = []   # full sequences for every (sample, choice)
    all_images  = []   # repeated image per candidate
    sample_idxs = []   # which sample each row belongs to

    for s_idx, (prompt, choices, img) in enumerate(zip(prompt_texts, choices_list, images)):
        for choice_text in choices:
            # CODE CHANGE: append the full choice TEXT instead of a single letter.
            # e.g. prompt + "West Virginia" instead of prompt + " A"
            full_seq = prompt + choice_text
            all_texts.append(full_seq)
            all_images.append(img)
            sample_idxs.append(s_idx)

    # Tokenise all candidates in one batch
    enc = processor(
        text=all_texts,
        images=all_images,
        return_tensors="pt",
        padding=True,
        truncation=True,
        max_length=MAX_SEQ_LEN,
    )
    enc = {k: v.to(model.device) if torch.is_tensor(v) else v for k, v in enc.items()}

    # Tokenise just the prompts to find the answer text start position
    prompt_enc = processor(
        text=[prompt_texts[s_idx] for s_idx in sample_idxs],
        images=all_images,
        return_tensors="pt",
        padding=True,
        truncation=True,
        max_length=MAX_SEQ_LEN,
    )
    prompt_lens = prompt_enc["attention_mask"].sum(dim=1)  # (N_cands,)

    # Forward pass
    with torch.amp.autocast("cuda", dtype=amp_autocast_dtype, enabled=torch.cuda.is_available()):
        logits = model(**enc).logits   # (N_cands, L, V)

    log_probs = F.log_softmax(logits, dim=-1)   # (N_cands, L, V)
    input_ids = enc["input_ids"]                # (N_cands, L)
    N         = input_ids.shape[0]

    cand_scores = torch.zeros(N, device=logits.device)
    for n in range(N):
        plen    = int(prompt_lens[n].item())
        seq_len = int(enc["attention_mask"][n].sum().item())

        # Number of answer-text tokens (excludes prompt and padding)
        # CODE CHANGE: used for length normalisation below
        num_answer_tokens = seq_len - plen

        if num_answer_tokens <= 0:
            continue  # degenerate case: answer text was truncated entirely

        # Sum log-probs over answer-text positions
        # (logits[t] predicts token[t+1], so we index from plen-1)
        for t in range(plen - 1, seq_len - 1):
            next_token = input_ids[n, t + 1]
            cand_scores[n] += log_probs[n, t, next_token]

        # CODE CHANGE: divide by token count to get per-token average log-prob.
        # Without this, longer choices accumulate more probability mass and
        # would be systematically preferred over shorter ones.
        cand_scores[n] /= num_answer_tokens

    # Argmax per sample
    B     = len(prompt_texts)
    preds = []
    for s_idx in range(B):
        mask = [i for i, si in enumerate(sample_idxs) if si == s_idx]
        best = int(torch.argmax(cand_scores[mask]).item())
        preds.append(best)

    return preds


print("Multiple-choice scoring function defined (full text + length normalisation).")
```

## 6. Training Loop

```python
# ── 6a. Optimizer & Scheduler ─────────────────────────────────────────────────
optimizer = torch.optim.AdamW(
    filter(lambda p: p.requires_grad, model.parameters()),
    lr=LR,
    weight_decay=0.01,
)

total_steps   = math.ceil(len(train_loader) / GRAD_ACCUM) * EPOCHS
warmup_steps  = int(total_steps * WARMUP_RATIO)

scheduler = get_cosine_schedule_with_warmup(
    optimizer,
    num_warmup_steps=warmup_steps,
    num_training_steps=total_steps,
)

print(f"Total optimizer steps : {total_steps:,}")
print(f"Warmup steps          : {warmup_steps:,}")
```

```python
# ── 6b. Helper: validate on the entire val set ────────────────────────────────
def run_validation(model, val_loader) -> tuple[float, float]:
    """
    Returns (val_loss, val_accuracy).
    val_loss     — average NLL of the correct answer TEXT over the val set.
    val_accuracy — fraction of questions answered correctly via MC scoring.
    """
    model.eval()
    total_loss  = 0.0
    total_steps = 0
    correct     = 0
    total       = 0

    for batch in val_loader:
        images   = batch["images"]
        texts    = batch["texts"]
        choices  = batch["choices"]
        answers  = batch["answers"]

        # MC accuracy — score_choices_batch already uses full text + length norm
        preds = score_choices_batch(model, processor, images, texts, choices)
        for pred, gt in zip(preds, answers):
            correct += int(pred == gt)
            total   += 1

        # ── Val loss (NLL on correct answer TEXT) ─────────────────────────────
        # CODE CHANGE: append the full correct choice text (not the letter) to
        # build the ground-truth sequence for loss computation. This keeps val
        # loss consistent with the training objective.
        eos = processor.tokenizer.eos_token
        gt_texts = [
            texts[i] + choices[i][answers[i]] + eos
            for i in range(len(texts))
        ]
        enc = processor(
            text=gt_texts,
            images=images,
            return_tensors="pt",
            padding=True,
            truncation=True,
            max_length=MAX_SEQ_LEN,
        )
        enc = {k: v.to(model.device) if torch.is_tensor(v) else v for k, v in enc.items()}

        prompt_enc = processor(
            text=texts,
            images=images,
            return_tensors="pt",
            padding=True,
            truncation=True,
            max_length=MAX_SEQ_LEN,
        )
        prompt_lens = prompt_enc["attention_mask"].sum(dim=1)

        labels = enc["input_ids"].clone()
        for i, plen in enumerate(prompt_lens):
            labels[i, :plen] = -100
        labels[enc["attention_mask"] == 0] = -100
        enc["labels"] = labels

        with torch.inference_mode():
            with torch.amp.autocast("cuda", dtype=amp_autocast_dtype, enabled=torch.cuda.is_available()):
                loss = model(**enc).loss
        total_loss  += loss.item()
        total_steps += 1

    val_loss = total_loss / max(total_steps, 1)
    val_acc  = correct / max(total, 1)
    return val_loss, val_acc
```

```python
# ── 6c. Main training loop ────────────────────────────────────────────────────
scaler = torch.amp.GradScaler('cuda', enabled=torch.cuda.is_available() and amp_autocast_dtype == torch.float16)

train_metrics = []   # will be saved to disk at end
best_val_acc  = -1.0
global_step   = 0
train_start   = time.time()   # ← total training timer

EARLY_STOP_PATIENCE  = 2
epochs_no_improve    = 0

for epoch in range(1, EPOCHS + 1):
    model.train()
    epoch_loss     = 0.0
    epoch_steps    = 0
    optimizer.zero_grad()

    for batch_idx, batch_i in enumerate(train_loader):
        # Move tensors to device
        batch = {k: v.to(model.device, non_blocking=True) if torch.is_tensor(v) else v
                 for k, v in batch_i.items()}
        # Explicitly cast attention_mask to bool and ensure contiguity
        # if "attention_mask" in batch and batch["attention_mask"] is not None:
        #     batch["attention_mask"] = batch["attention_mask"].bool().contiguous()

        with torch.amp.autocast("cuda", dtype=amp_autocast_dtype, enabled=torch.cuda.is_available()):
        # with torch.amp.autocast("cuda", enabled=torch.cuda.is_available()):
            outputs = model(**batch)
            loss    = outputs.loss / GRAD_ACCUM

        # Only scale if float16 is used for autocasting
        if amp_autocast_dtype == torch.float16:
            scaler.scale(loss).backward()
        else:
            loss.backward()

        if (batch_idx + 1) % GRAD_ACCUM == 0 or (batch_idx + 1) == len(train_loader):
            if amp_autocast_dtype == torch.float16:
                scaler.unscale_(optimizer)
            torch.nn.utils.clip_grad_norm_(
                filter(lambda p: p.requires_grad, model.parameters()), 1.0
            )
            if amp_autocast_dtype == torch.float16:
                scaler.step(optimizer)
                scaler.update()
            else:
                optimizer.step()
            scheduler.step()
            optimizer.zero_grad()
            global_step += 1

        epoch_loss  += loss.item() * GRAD_ACCUM   # unscale for logging
        epoch_steps += 1

        if (batch_idx + 1) % 100 == 0:
            avg = epoch_loss / epoch_steps
            lr_now = scheduler.get_last_lr()[0]
            print(f"  Epoch {epoch} | step {batch_idx+1}/{len(train_loader)} "
                  f"| loss {avg:.4f} | lr {lr_now:.2e}")

    avg_train_loss = epoch_loss / epoch_steps

    # ── End-of-epoch validation ───────────────────────────────────────────────
    print(f"\nEpoch {epoch} complete — running validation…")
    val_loss, val_acc = run_validation(model, val_loader)

    epoch_record = {
        "epoch":          epoch,
        "train_loss":     round(avg_train_loss, 6),
        "val_loss":       round(val_loss,       6),
        "val_accuracy":   round(val_acc,        6),
        "global_step":    global_step,
        "lr":             scheduler.get_last_lr()[0],
        "epoch_time_sec": round(time.time() - train_start, 1),
    }
    train_metrics.append(epoch_record)

    print(f"\n{'='*60}")
    print(f"Epoch {epoch:2d} | train_loss={avg_train_loss:.4f} "
          f"| val_loss={val_loss:.4f} | val_acc={val_acc:.4f}")
    print(f"{'='*60}\n")

    # ── Save best checkpoint ──────────────────────────────────────────────────
    if val_acc > best_val_acc:
        best_val_acc = val_acc
        epochs_no_improve = 0   # IMPROVEMENT 8: reset patience counter
        model.save_pretrained("best_checkpoint")
        processor.save_pretrained("best_checkpoint")
        print(f"  ✓ New best model saved (val_acc={best_val_acc:.4f})")
    else:
        epochs_no_improve += 1  # IMPROVEMENT 8: increment patience counter
        print(f"  No improvement ({epochs_no_improve}/{EARLY_STOP_PATIENCE} patience epochs used)")

    # IMPROVEMENT 8: halt if patience exhausted
    if epochs_no_improve >= EARLY_STOP_PATIENCE:
        print(f"\nEarly stopping triggered after {epoch} epochs (no val_acc improvement "
              f"for {EARLY_STOP_PATIENCE} consecutive epochs).")
        break

# ── Save training metrics to disk ─────────────────────────────────────────────
total_training_time = round(time.time() - train_start, 1)
val_wrong_answers   = round((1 - best_val_acc) * len(val_df))
with open(DRIVE_DIR_PATH + f"/train_metrics{SUB_RUN_ID}.json", "w") as f:
    json.dump(
        {
            "config":               cfg,
            "best_val_acc":         best_val_acc,
            "total_training_time_sec": total_training_time,
            "val_wrong_answers":    val_wrong_answers,
            "epochs":               train_metrics,
        },
        f,
        indent=2,
    )

print(f"\nTraining complete in {total_training_time:.1f}s — metrics saved to train_metrics.json")
```
